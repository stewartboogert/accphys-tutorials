# Energy gain

The work done $\Delta W$ on a charged particle is ${\bf F}\cdot d{\bf s}$, integrating over a path where there is a field

$$
\begin{eqnarray}
\Delta W & = & \int_{s_0}^{s_1} {\bf F} \cdot d{\bf s} \\
         & = & \int_{s_0}^{s_1} q {\bf E} \cdot d{\bf s}
\end{eqnarray}
$$

The field typically is radio-frequency RF i.e oscillating at an angular frequency $\omega_{\rm RF}$, so assuming it is just a function of $z$ so   

$$
\begin{eqnarray}
\Delta W & = & q \int_{z_0}^{z_1} E(r=0,z) \exp (i \omega_{\rm RF} t + \phi) dz \\
         & = & q \int_{z_0}^{z_1} E(r=0,z) \cos ( \omega_{\rm RF} t + \phi) dz \\
         & = & q \int_{z_0}^{z_1} E(r=0,z) \left[ \cos ( \omega_{\rm RF} t ) \cos (\phi) - \sin(\omega_{\rm RF}t)\sin(\phi) \right] dz \\
         & = & q \cos (\phi) \int_{z_0}^{z_1} E(r=0,z) \cos ( \omega_{\rm RF} t ) dz   - \sin(\phi)  \int_{z_0}^{z_1} E(r=0,z) \sin(\omega_{\rm RF}t) dz \\
         & = & q \cos (\phi)\int_{z_0}^{z_1} E_{r=0,z} dz \left[ \frac{\int_{z_0}^{z_1} E(r=0,z) \cos ( \omega_{\rm RF} t ) dz}{\int_{z_0}^{z_1} E(r=0,z) dz} -\tan \phi \frac{\int_{z_0}^{z_1} E(r=0,z) \sin ( \omega_{\rm RF} t ) dz}{\int_{z_0}^{z_1} E(r=0,z) dz}\right] \\
         & = & q V_0 T \cos \phi
\end{eqnarray}
$$

where we have defined the voltage $V_0$ as  

$$
V_0 = \int_{z_0}^{z_1} E(r=0,z) \;dz
$$

and $T$, the transit time factor as 

$$
T =  \frac{\int_{z_0}^{z_1} E(r=0,z) \cos ( \omega_{\rm RF} t ) dz}{\int_{z_0}^{z_1} E_{r=0,z} dz} -\tan \phi \frac{\int_{z_0}^{z_1} E(r=0,z) \sin ( \omega_{\rm RF} t ) dz}{\int_{z_0}^{z_1} E_{r=0,z} dz}
$$

If the arrival phase is zero $\phi =0$ so

$$
T = \frac{\int_{z_0}^{z_1} E(r=0,z) \cos ( \omega_{\rm RF} t ) dz}{\int_{z_0}^{z_1} E_{r=0,z} dz}
$$

We now need to convert the time argument of the trigonometric function to distance. The wavelength of a oscillation of frequency $\omega_{\rm RF}$ is easily calculated as 

How far does a particle travel in one RF period $\tau$

$$
\begin{eqnarray}
\tau & = & \frac{1}{f} \\
     & = & \frac{2\pi}{\omega}
\end{eqnarray}
$$

$z_{\rm RF}$ is the distance travelled in one period
$$
\begin{eqnarray}
z_{\rm RF} & = & v \tau \\
           & = & \beta c \frac{2\pi}{\omega} \\ 
           & = & \beta \lambda
\end{eqnarray}
$$
where $\lambda$ is the wavelength of the RF ocillation. So argument of the cosine is 

$$
\cos (\omega_{\rm RF} t) = \cos \left(2\pi \frac{z}{\beta \lambda} \right)
$$

So the transit time factor becomes 

$$
T = \frac{\int_{z_0}^{z_1} E(r=0,z) \cos \left(2\pi \frac{z}{\beta \lambda} \right) dz}{\int_{z_0}^{z_1} E(r=0,z) dz}
$$

If $E(r=0,z)$ is constant along $z$, i.e $E(r=0,z) = E_0$, then the transit time factor is 

$$
T = \frac{\int_{z_0}^{z_1} \cos \left(2\pi \frac{z}{\beta \lambda} \right) dz}{z_1 - z_0}
$$

Then integrating (assuming the gap is symmetric $z_1 = g/2$ and $z_0=-g/2$ 

$$
\begin{eqnarray}
T & = & \frac{1}{z_1 - z_0}\left[\frac{\sin \left(2\pi \frac{z}{\beta \lambda} \right)} {\frac{2\pi}{\beta \lambda}}\right]_{z_0}^{z_1} \\
  & = & \frac{1}{\frac{2 \pi g}{\beta \lambda}} \left[ \sin \left(2\pi \frac{z_1}{\beta \lambda} \right) - \sin \left(2\pi \frac{z_0}{\beta \lambda} \right)  \right] \\
  & = & \frac{1}{\frac{2 \pi g}{\beta \lambda}} \left[ \sin \left(2\pi \frac{g/2}{\beta \lambda} \right) - \sin \left(-2\pi \frac{g/2}{\beta \lambda} \right)  \right] \\
  & = & \frac{\sin \left(\frac{\pi g}{\beta \lambda} \right)}{ \frac{\pi g}{\beta \lambda}}
\end{eqnarray}
$$
