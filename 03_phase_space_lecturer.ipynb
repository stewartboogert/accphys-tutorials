```python
import numpy as np

def gam(alp, bet) :
    return (1+alp**2)/bet

def one_turn(alp, bet, gam, mu) :
    return np.array([[np.cos(mu) + alp*np.sin(mu), bet*np.sin(mu)],
                      [-gam*np.sin(mu), np.cos(mu) - alp*np.sin(mu)]])
    
```

```python
def phase_space(alp=0.2 , bet=1 , mu=0.2, x = 1.0 ,xp = 0.0, npoint = 1000) :
    mot = one_turn(alp,bet,gam(alp,bet), 0.2)
    v = np.array([x,xp])
    vlist = []
    for i in range(npoint) :
        v = mot @ v
        vlist.append(v)
    varray = np.array(vlist)
    return varray
```

```python
import matplotlib.pyplot as plt 
varray = phase_space(alp=0.2, bet=1, mu=0.2, x=1, xp=0)
plt.plot(varray[:,0], varray[:,1])
plt.xlabel("x");
plt.ylabel("xp");
```

```python
from ipywidgets import interact

def plot(alp = 0, bet= 1, mu=0.2, npoint=100) :
    varray = phase_space(alp=alp, bet=bet, mu=mu, x=1, xp=0, npoint=npoint)
    plt.plot(varray[:,0], varray[:,1],"o")
    plt.xlabel("x");
    plt.ylabel("xp");

interact(plot, alp=(-10, 10.0, 1.0), bet=(0,10,1.0), npoint=(1,500,1), mu=(0,2*np.pi,0.1));
```
