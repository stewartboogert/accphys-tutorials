# Transverse dynamics

In lectures we found that the effect of some magnetic elements can be described as matricies ${\bf M}$ which act on a vector of coordinates ${\bf v}$, going from one location labelled 1 to another labelled 2 gives  

$$
\begin{equation}
{\bf v}_1 = {\bf M} {\bf v}_0
\end{equation}
$$

To explore these ideas we will use matricies and vectors from numpy. So we will need to import the numpy module  

```python
import numpy as np
```

As a quick refresher we will need to create vectors and matrices and perform matrix-vector multiplication.

$$
\begin{equation}
{\bf v}_0 = \begin{bmatrix} x_0\\ x_0^{\prime} \end{bmatrix}
\end{equation}
$$

To create this vector in numpy with $x_0 = 0$ and $x_0^{\prime} = 1$ 

```python
v0 = np.array([0,-1])
```

The simplest transfer matrix is that of a drift through as distance of $L$, so 

$$
{\bf M}_{\rm drift} = \begin{bmatrix} 1 & L \\ 0 & 1 \end{bmatrix}
$$

To create this $2\times 2$ matrix in numpy, with a length of 5 m

```python
m = np.array([[1, 5], [0, 1]])
```

Now to see what happens to the coordinates of a particle, by applying ${\bf M}_{\rm drift}$ to ${\bf v}_0$. For numpy to perform matrix multiplication the `@` operator and not `*` is used

```python
m @ v0
```

## Example : multiple drifts

If we had multiple different accelerator elements, to calculate the transfer matrix they need to be applied in beam line order from start to end, so if there are three elements, labelled 1, 2 and 3 with matricies ${\bf M}_1$, ${\bf M}_2$ and ${\bf M}_3$ respectively we would have 

$$
{\bf v}_3 = {\bf M}_1 {\bf M}_2 {\bf M}_3 {\bf v}_0
$$
