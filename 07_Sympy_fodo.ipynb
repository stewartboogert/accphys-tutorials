# Focusing-defocising lattice

Drift transfer matrix
$$
D(L) = \begin{bmatrix} 1 & L \\ 0 & 1 \end{bmatrix}  
$$

Focusing "thin" quadrupole
$$
M_{F} = \begin{bmatrix} 1 & 0 \\ -\frac{1}{f_F} & 1 \end{bmatrix} 
      = \begin{bmatrix} 1 & 0 \\ -K_F L_F & 1 \end{bmatrix}
$$

Defocusing "thin" quadrupole
$$
M_{D} = \begin{bmatrix} 1 & 0 \\ \frac{1}{f_D} & 1 \end{bmatrix}
      = \begin{bmatrix} 1 & 0 \\ K_D L_D & 1 \end{bmatrix}
$$

Half focusing, drift, defocusing, drift, half focusing

$$
{\bf M}=
\begin{bmatrix} 1 & 0 \\ -\frac{1}{2 f} & 1 \end{bmatrix} 
\begin{bmatrix} 1 & L \\ 0 & 1 \end{bmatrix} 
\begin{bmatrix} 1 & 0 \\ \frac{1}{f} & 1 \end{bmatrix} 
\begin{bmatrix} 1 & L \\ 0 & 1 \end{bmatrix}
\begin{bmatrix} 1 & 0 \\ -\frac{1}{2 f} & 1 \end{bmatrix} = \begin{bmatrix}- \frac{L^{2}}{2 f^{2}} + 1 & \frac{L \left(L + 2 f\right)}{f} \\ \frac{L \left(L - 2 f\right)}{4 f^{3}} & - \frac{L^{2}}{2 f^{2}} + 1\end{bmatrix}
$$

Drift, focusing, drift, defocusing

$$
{\bf M}= 
\begin{bmatrix} 1 & L \\ 0 & 1 \end{bmatrix} 
\begin{bmatrix} 1 & 0 \\ -\frac{1}{f} & 1 \end{bmatrix} 
\begin{bmatrix} 1 & L \\ 0 & 1 \end{bmatrix} 
\begin{bmatrix} 1 & 0 \\ \frac{1}{f} & 1 \end{bmatrix} = \begin{bmatrix} -\frac{L^{2}}{f^{2}} + \frac{L}{f} + 1 & \frac{L \left(- L + 2 f\right)}{f}\\- \frac{L}{f^{2}} & \frac{- L + f}{f}\end{bmatrix}
$$

So how to do you verify if the matrix is stable. How about keep on applying a matrix ${\bf M}^N{\bf v}$. We can look at the Eigenvectors of the matrix ${\bf M}$
#
$$
\begin{eqnarray}
{\bf M} {\bf v} & = & \lambda {\bf v}
\end{eqnarray}
$$
What if one of the Eigenvalues $\lambda$ is larger than 1, $\lambda > 1$. If we keep on applying the matrix ${\bf M}$ the phase vector is going to get larger and larger in magnitude.  


```python
import sympy as sp

L = sp.symbols('L', positive=True)
D = sp.Matrix([[1, L],
               [0, 1]])

f = sp.symbols('f', positive=True)
MF_half = sp.Matrix([[1, 0],
               [-1/(2*f), 1]])

MF = sp.Matrix([[1, 0],
               [-1/f, 1]])


MD = sp.Matrix([[1, 0],
               [1/f, 1]])
```

## Half focusing quad, drift, Defocusing quad, drift, half focusing quad

```python
M_QDQDQ = MF_half @ D @ MD @ D @ MF_half
```

```python
M_QDQDQ.simplify()
```

```python
M_QDQDQ
```

## Drift, focusing quad, drift, defocusing quad

```python
M_DQDQ = D @ MF @ D @ MD 
```

```python
M_DQDQ.simplify()
```

```python
M_DQDQ
```

## Stability

$$
\cos \mu = \frac{1}{2}{\rm Tr}({\bf M})
$$

```python
M_QDQDQ.trace()
```

```python
M_DQDQ.trace().simplify()
```

Using the stability condition

$$
\cos \mu = 1-\frac{L^2}{2f^2}
