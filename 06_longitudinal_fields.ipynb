# Electromagnetic fields

In general only the electric field can do work, so taking the Lorentz force and integrating over a path

$$
W = q \int \left( {\bf E} + {\bf v} \times {\bf B}  \right) \cdot d{\bf r},
$$
the second term is always perpendicular to the displacement.

So in general we would like the electric field to point along the direction of travel. This would make the first term ${\bf E} \cdot d{\bf r}$ maximum. So an accelerator would like to have as large as possible electric field in the direction of the particles velocity.

A simple way of doing this is a parallel plate capacitor. So two plates with a voltage between them, of course with a hole for the beam to go through.

This is done in some accelerators, but in general is avoided as the large field is not needed all the time. In general a radio-frequency (RF) field is used. For accelerators there are two main oscillating field configurations. Called the **transverse magnetic** TM and **transverse electric** TE. In an electromagnetic plane wave is a **transverse electromagnetic** TEM mode.
