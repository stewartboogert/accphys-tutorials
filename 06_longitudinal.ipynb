# Longitudinal dynamics

## Synchronous particle and energy gain

The revolution period is related to the circumferance $C$ and velocity $v$ by 

$$
T_{\rm rev} = \frac{C}{v} = \frac{2\pi R}{\beta c},
$$
were $R$ is a machine radius, $\beta$ is relativistic velocity and $c$ is the speed of light.

The angular frequency of revolution $\omega_{\rm rev}$ is 

$$
\omega_{\rm rev} = 2\pi f_{\rm rev} = \frac{2\pi}{T_{\rm rev}} = \frac{\beta c}{R}
$$

A particle is synchronous with the RF if 

$$
\omega_{\rm rf} = h \omega_{\rm rev,s} = h \frac{\beta_s c}{R_s},
$$
where $h$ is an integer, $\beta_s$ and $R_{s}$ refer to the synchronous velocity and machine radius.

The energy gain per turn is given by 

$$
\delta E =q V_{\rm rf} \sin(\phi_s),
$$
where $\phi_s$ is the synchronous phase.

The rate of change of energy is easily calculated as we know how long a revolution takes 

$$ 
\frac{dE_{s}}{dt} \approx \frac{\delta E}{T_{\rm rev,s}} = \frac{q V_{\rm rf}}{T_{\rm rev,s}} \sin(\phi_s)
$$

The magnetic field needs to vary at the same rate to keep the particle synchronous and on the reference orbit. Recalling 

$$
B \rho_s = \frac{p}{q},
$$
so the time derivative is 

$$
\frac{dB}{dt} \rho_s = \frac{1}{q}\frac{dp}{dt}
$$

Recalling the energy $E$ and momentum $p$ are related by $dE/dp = v = \beta c$, so 

$$
\begin{eqnarray}
\delta E_{s} & = & \frac{dE_{s}}{dt} T_{\rm rev,s} \\
             & = & q \rho_s \beta c \frac{dB}{dt} T_{\rm rev,s} \\
             & = & q \rho_s \beta c \frac{dB}{dt} \frac{2\pi R_{s}}{\beta c} \\
             & = & 2\pi q R_{s} \rho_s \frac{dB}{dt}
\end{eqnarray}
$$

so the rate of change of the magnetic field has to linked to the energy gain.
