# Matching from one set of Twiss values to another 

At some location $s_0$ we have Twiss parameters $(\beta, \alpha, \gamma)_0$ and would would like to manipulate these using magnets to a target set of values $(\beta, \alpha, \gamma)_1$. 

We could image doing this with a set of drifts and quads $(D_1, Q_1, D_2, Q_2, D_3)$

```python
import sympy as sp

def drift_matrix(L) :
    return sp.Matrix([[1, L],[0,1 ]])

def quad_matrix(f) :
    return sp.Matrix([[1, 0],[1/f,1 ]])

def twiss_propagation(m) :
    return sp.Matrix([[m[0,0]**2, -2*m[0,0]*m[0,1], m[0,1]**2],
                      [-m[0,0]*m[1,0], 1+ 2*m[0,1]*m[1,0], -m[0,1]*m[1,1]],
                      [m[1,0]**2, -2*m[1,0]*m[1,1], m[1,1]**2]])

# make the drifts 
l1 = sp.symbols('l1',positive=True)
l2 = sp.symbols('l2',positive=True)
l3 = sp.symbols('l3',positive=True)
d1m = drift_matrix(l1)
d2m = drift_matrix(l2)
d3m = drift_matrix(l3)

# make the quads 
f1 = sp.symbols('f_1')
f2 = sp.symbols('f_2')
q1m = quad_matrix(f1)
q2m = quad_matrix(f2)

MATCH = d3m @ q2m @ d2m @ q1m @ d1m
```

```python
MATCH
```

```python
MATCH_TWISS = twiss_propagation(MATCH)
```

```python
MATCH_TWISS.simplify()
```
