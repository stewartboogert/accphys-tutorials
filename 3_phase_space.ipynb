# Phase ellipse 

Typically we represent the $(x, x^{\prime})$ coordinates of a particle in a bunch on a plot. All the points are distributed as a 2-D Gaussian in which $x$ and $x^{\prime}$ are correlated, this then looks like an ellipse. In a circular accelerator all the transfer matricies can be combined to form a single matrix, this is called the one turn transfer matrix, so 

$$
\begin{equation}
{\bf M}_{\rm one turn} = {\bf M}_1 {\bf M}_2 {\bf M}_3 ... {\bf M}_N  
\end{equation}
$$

In general this matrix can be written in the (Courant-Snyder) form 

$$
\begin{equation}
{\bf M}_{\rm one turn} = \begin{bmatrix} 
            \cos \mu + \alpha \sin \mu & \beta \sin \mu \\
            -\gamma \sin \mu           & \cos \mu - \alpha \sin \mu
           \end{bmatrix} 
\end{equation}
$$

The parameters $\alpha$, $\beta$ and $\gamma$ are the Twiss parameters we learned about in lectures. All three are not independent, only 2 of the 3 are. As a reminder 

$$
\begin{equation}
\alpha(s) = - \frac{1}{2} \frac{d\beta(s)}{ds}
\end{equation}
$$

and 

$$
\gamma = \frac{1+\alpha(s)^2}{\beta(s)}.
$$
